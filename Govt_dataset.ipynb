# Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path =r"C:\Users\Aswini\Downloads\Atal_Jal_Area_0.csv"
df = pd.read_csv(file_path)
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(y=df["state_name"], order=df["state_name"].value_counts().index)
plt.title("Number of Entries per State")
plt.xlabel("Count")
plt.ylabel("State Name")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(y=df["district_name"], order=df["district_name"].value_counts().index[:10])
plt.title("Top 10 Districts with Most Entries")
plt.xlabel("Count")
plt.ylabel("District Name")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
df["district_name"].value_counts().head(10).plot(kind="bar", color="skyblue")
plt.title("Top 10 Districts with Most Entries")
plt.xlabel("District Name")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
df["block_name"].value_counts().plot(kind="hist", bins=20, color="orange", edgecolor="black")
plt.title("Distribution of Entries per Block")
plt.xlabel("Number of Entries")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 8))
df["state_name"].value_counts().plot(kind="pie", autopct='%1.1f%%', startangle=90, cmap='viridis')
plt.title("State-wise Distribution of Entries")
plt.ylabel("")  
plt.show()
```

```python
plt.figure(figsize=(10, 5))
df["block_name"].value_counts().sort_index().plot(kind="line", marker="o", color="green")
plt.title("Trend of Entries per Block")
plt.xlabel("Block Name")
plt.ylabel("Number of Entries")
plt.xticks(rotation=90)
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(15, 7))
district_counts = df.groupby("state_name")["district_name"].nunique().sort_values(ascending=False)
district_counts.plot(kind="bar", color=["purple", "blue", "green", "red", "orange"])
plt.title("Number of Districts per State")
plt.xlabel("State Name")
plt.ylabel("Number of Districts")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df.select_dtypes(include=['number']))
plt.title("Box Plot for Numeric Columns")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
if df.select_dtypes(include=['number']).shape[1] >= 2:
    numeric_cols = df.select_dtypes(include=['number']).columns[:2]
    sns.lmplot(x=numeric_cols[0], y=numeric_cols[1], data=df, aspect=1.5, scatter_kws={'alpha':0.5})
    plt.title("Linear Relationship between {} and {}".format(numeric_cols[0], numeric_cols[1]))
else:
    print("Not enough numeric columns for a linear plot.")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
if df.select_dtypes(include=['number']).shape[1] >= 2:
    numeric_cols = df.select_dtypes(include=['number']).columns[:2]
    plt.scatter(df[numeric_cols[0]], df[numeric_cols[1]], alpha=0.5, color='blue')
    plt.title("Scatter Plot between {} and {}".format(numeric_cols[0], numeric_cols[1]))
    plt.xlabel(numeric_cols[0])
    plt.ylabel(numeric_cols[1])
    plt.grid()
else:
    print("Not enough numeric columns for a scatter plot.")
plt.show()
```
